```python
def square_wave(n, f, phi):
    return np.sign(np.sin(tau * f * n + phi))
```

```python
plot(square_wave(np.arange(50), 0.1, 0))
```

```python
def dv_square_wave(n, f, phi):
    delta_f = 1e-8
    return (square_wave(n, f + delta_f, phi) - square_wave(n, f - delta_f, phi)) / (2 * delta_f)
```

```python
f_axis = np.linspace(0, 0.5, 1000)
```

```python
fi = [(dv_square_wave(np.arange(50), f)**2).sum() for f in f_axis]
fi = np.array(fi)
```

```python
plot(f_axis, fi)
scatter(np.array([0.1, 0.125, 0.15625, 0.2, 0.22, 0.25, 0.3125, 0.34375, 0.4]), [0]*9, c='r')
```
