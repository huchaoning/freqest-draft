```python
import matplotlib.pyplot as plt 
import numpy as np 
from scipy.integrate import quad

from matplotlib_inline.backend_inline import set_matplotlib_formats
set_matplotlib_formats('retina')


from matplotlib.font_manager import fontManager
fontManager.addfont('/Users/me/Desktop/freqest/src/lmroman10.otf')
del fontManager

plt.rcParams['font.family'] = 'Latin Modern Roman'
plt.rcParams['mathtext.fontset'] = 'cm'
```

```python
def i(b, sigma=1):
    b_ = b * sigma
    func = lambda x: (x**2 * np.exp(-x**2)) / (np.sqrt(2*np.pi)*np.exp(-x**2/2) + 2*np.pi*b_)
    return quad(func, -np.inf, np.inf)[0] / sigma**2
```

```python
x = np.linspace(10e-10, 1, 1000)
y = np.array([i(x_) for x_ in x])
```

```python
plt.rcParams['font.size'] = 7
plt.rcParams['savefig.bbox'] = 'tight'

plt.figure(figsize=(8.5 / 2.54, 8.5 / 2.54))

plt.plot(x, y, 'b')
plt.grid(False)
plt.xlim(0, 1)
plt.ylim(0, 1)
plt.xlabel(r"$b' \sigma$")
plt.ylabel(r'$\sigma^2\gamma^\text{(DI)}$')

plt.gca().set_box_aspect(0.618)

plt.savefig('/Users/me/Desktop/N_int.pdf')
plt.show()
```
