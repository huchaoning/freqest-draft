```python
freq_list = np.linspace(0.1, 0.4, 61)
```

```python
for _f in tqdm(freq_list):
    f = np.round(_f, 5)

    raw = np.load(rf'__raw__/spade_5px/spade_{f}_raw.npy')
    timestamp = np.load(rf'__raw__/spade_5px/spade_{f}_timestamp.npy')

    meta = MetaData('SPADE', f, 5*DMD.PIXEL_SIZE/2, timestamp)

    c = FrequencyEstimation.FromRaw(raw, meta)
    c.savez(rf'__estimates__/spade_5px/spade_{f}.npz')
    gc.collect()
```
