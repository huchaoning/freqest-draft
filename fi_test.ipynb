```python
def fim(A, freq, phi):
    n = np.arange(50)

    part_a = -1 + np.sin(tau*freq*n + phi)
    part_b = 2*A*n*pi*np.cos(tau*freq*n + phi)
    part_c = A*np.cos(tau*freq*n + phi)

    return np.array([[part_a * part_a, part_a * part_b, part_a * part_c],
                     [part_b * part_a, part_b * part_b, part_b * part_c], 
                     [part_c * part_a, part_c * part_b, part_c * part_c]]).sum(-1) / 103**2
```

```python
x = np.linspace(0.1, 0.4, 1000)

plot(x, [np.linalg.inv(fim(5 * DMD.PIXEL_SIZE / 2, f, 0))[1, 1] for f in x], label=r'$A\sin (2\pi f n +\phi) - A$')

plot(x, [np.linalg.inv(FIM_CRB(phi=0.416).fim(5 * DMD.PIXEL_SIZE / 2, f))[0, 0] for f in x], label=r'$\sin (2\pi f n +\phi)$')

plot(x, [np.linalg.inv(FIM_CRB(approx=True).fim(5 * DMD.PIXEL_SIZE / 2, f))[0, 0] for f in x], label=r'$\sin (2\pi f n +\phi)$')

plot(title='CRB', xlabel='Frequency', ylabel='Var')
```
