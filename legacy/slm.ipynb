```python
from myutils import *
from scipy.stats import norm
```

```python
def fftimg(U):
    W = np.fft.fftshift(np.fft.fft2(U))
    return np.abs(W)**2
```

```python
x, y = fast_meshgrid(1024, 1024, 0.01)
```

```python
optical_field = norm(0, 1).pdf(x) * norm(0, 1).pdf(y)
```

```python
imshow(fftimg(optical_field))
```
