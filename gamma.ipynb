```python
b_range = np.linspace(0, 0.1, 300)

A = 0.47*103

QCRB = DI.CFI(0, A, 0.2)

plt.plot(b_range, [DI.CFI(b, A, 0.2) / QCRB for b in b_range], label='Direct imaging')
plt.plot(b_range, [PM_SPADE.CFI(b, A, 0.2) / QCRB for b in b_range], label='PM-SPADE')

plt.plot(b_range, [HG_SPADE.CFI(b, A, 0.2) / QCRB for b in b_range], label='HG-SPADE')
plt.xlabel(r'$b / \nu$')
plt.ylabel(r'$F / \mathcal{F}$')
plt.ylim(0, 1)
plt.xlim(0, 0.1)
plt.title(r'$A=0.47\sigma;\quad f=0.2$')
plt.legend(frameon=False)
plt.grid(False)

plt.savefig('small_b.svg')
plt.show()
```
